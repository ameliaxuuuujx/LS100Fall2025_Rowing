# Mixed-Effect Model
## Mathematical Formula
\begin{equation}
s_{Q,ij} = \beta_0 + \beta_1 \text{StrokeRate}{ij} + \beta_2 \text{SessionLength}{ij} + u_{0j} + \epsilon_{ij}
\end{equation}
- $i$: stroke (observation)
- $j$: participant
- $u_{0j}$: the random intercept for participant $j$. Some rowers may have a generally higher or lower technique quality score due to differences in overall skill level or consistency.
- $\epsilon_{ij}$: the residual error for each stroke. Represents within-participant variability (differences between individual strokes by the same person).
